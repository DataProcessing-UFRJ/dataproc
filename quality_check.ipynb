#### Quality accessment for all images in a folder (night)
It is probably a good idea to move only the stacked images (in which photometry will be carried out) into a different folder, to avoid wasting time getting quality accessment for images that will not be used in the photometry (like the individual images in each filter).


```python
from packages.dataprocessing_functions import fwhm_estimate

folder = './data/2025-11-18/phot'

fwhm_estimate(folder, fwhm_model='gaussian', redo_fwhm=True, save_table=True, 
              border_size=0.05, n_max=None, multiprocessing=True)
```
